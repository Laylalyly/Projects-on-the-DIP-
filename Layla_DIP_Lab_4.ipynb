```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('image.jpg')

kernel = np.ones((5,5), np.float32) / 25

smoothed = cv2.filter2D(img, -1, kernel)

def add_salt_and_pepper(image, ratio=0.05):
    noisy = np.copy(image)
    num_salt = np.ceil(ratio * image.size * 0.5)

    coords = [np.random.randint(0, i - 1, int(num_salt)) for i in image.shape]
    noisy[tuple(coords)] = 255

    coords = [np.random.randint(0, i - 1, int(num_salt)) for i in image.shape]
    noisy[tuple(coords)] = 0
    return noisy
    noisy_img = add_salt_and_pepper(img, 0.02)

def add_gaussian_noise(image, mean=0, sigma=25):
    gauss = np.random.normal(mean, sigma, image.shape).astype('uint8')
    noisy = cv2.add(image, gauss)
    return noisy
    gaussian_noisy = add_gaussian_noise(img, 0, 25)
    gaussian_blur = cv2.GaussianBlur(noisy_img, (5, 5), 0)
    median_blur = cv2.medianBlur(noisy_img, 5)


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  
plt.figure(figsize=(30, 6))
plt.subplot(1, 4, 1)
plt.imshow(img_rgb)
plt.title("Iameg")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(smoothed)
plt.title("Smoothed")
plt.axis("off")


plt.show()
```
