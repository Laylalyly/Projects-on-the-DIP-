```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img=cv2.imread('color.jpg',0)

sobelx_64 = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=5) 
sobely_64 = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5) 

sobelx=cv2.convertScaleAbs(sobelx_64)
sobely=cv2.convertScaleAbs(sobely_64)

sobel_combined=cv2.addWeighted(sobelx,0.5,sobely,0.5,0)



plt.subplot(1, 4, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(sobelx, cmap="gray")
plt.title("sobelx")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(sobely, cmap="gray")
plt.title("sobely")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(sobel_combined, cmap="gray")
plt.title("sobel_combined")
plt.axis("off")


plt.show()
```

```python
blurred = cv2.GaussianBlur(img, (3,3), 0)
laplacian_64 = cv2.Laplacian(blurred, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian_64)


plt.subplot(1, 3, 1)
plt.imshow(blurred, cmap="gray")
plt.title("blurred")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(laplacian, cmap="gray")
plt.title("laplacian")
plt.axis("off")

plt.show()
```

```python
edges = cv2.Canny(img, 100, 200)
cv2.imshow('Canny Edges', edges)

plt.subplot(1, 3, 1)
plt.imshow(edges, cmap="gray")
plt.title("edges")
plt.axis("off")

plt.show()
```

```python
def nothing (x):
    pass


    
cv2.namedWindow("Canny Edge Detector")


cv2.createTrackbar("Min Threshohd","Canny Edge Detector",
                   100 ,
                   255,
                   nothing)

cv2.createTrackbar("Max Threshohd","Canny Edge Detector",
                   200 ,
                   255,
                   nothing)


while True:
    minVal=cv2.getTrackbarPos("Min Threshohd","Canny Edge Detector")
    
    maxVal=cv2.getTrackbarPos("Max Threshohd","Canny Edge Detector")
    
    edges=cv2.Canny(img,minVal,maxVal)
    
    cv2.imshow("Canny Edge Detector ", edges)
    
    Key=cv2.waitKey(1)& 0xFF
    
    if Key== 27:
        break 
        
cv2.destroyAllWindows()
```
