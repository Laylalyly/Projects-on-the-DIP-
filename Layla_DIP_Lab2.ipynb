```python
import cv2
import numpy as np
img_bgr = cv2.imread('image.jpg')
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
print(img_bgr.shape)
```

```python
cv2.imshow('BGR Image', img_bgr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow('Gray Image', img_gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow('HSV Image', img_hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import matplotlib.pyplot as plt
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)  # للتحويل من BGR إلى RGB
plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")


plt.subplot(1, 3, 3)
plt.imshow(img_hsv)
plt.title("HSV")
plt.axis("off")

plt.show()
```

```python
apple_crop = img_rgb[200:750, 360:910]
print(img_rgb.shape)
```

```python
plt.imshow(apple_crop)
plt.title("Cropped RGB")
plt.axis("off")
plt.show()
```

```python
cv2.imshow('BGR Image', apple_crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
resized = cv2.resize(img_bgr, (224, 224), interpolation=cv2.INTER_LINEAR)
print(resized.shape)
```

```python
resized = cv2.resize(img_bgr, (224, 224), interpolation=cv2.INTER_NEAREST)
print(resized.shape)
```

```python
plt.imshow(resized)
plt.title("Resized RGB")
plt.axis("off")
plt.show()
```

```python
cv2.imshow('Resized Image', resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
(h, w) = img_bgr.shape[:2]
center = (w // 2, h // 2)

M = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated = cv2.warpAffine(img_bgr, M, (w, h))
print(rotated.shape)
print(center)
```

```python
plt.imshow(rotated)
plt.title("Rotated RGB")
plt.axis("off")
plt.show()
```

```python
cv2.imshow('Rotated Image', rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
